```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)

dates = pd.date_range(start="2026-01-01", periods=10, freq="D")

sp500 = np.array([
    6000, 6015, 5990, 6025, 6040,
    6030, 6060, 6075, 6055, 6090
])

vix = np.array([
    16.2, 15.8, 17.1, 16.4, 15.9,
    16.8, 15.5, 15.2, 16.1, 14.9
])

treasury_10y = np.array([
    4.20, 4.18, 4.22, 4.19, 4.17,
    4.21, 4.16, 4.15, 4.18, 4.14
])
```

```python
df = pd.DataFrame({
    "date": dates,
    "sp500": sp500,
    "vix": vix,
    "treasury_10y": treasury_10y
})

df.head()
```

```python
df["sp500_return"] = df["sp500"].pct_change()

df["vix_change"] = df["vix"].diff()

df["rolling_volatility"] = (
    df["sp500_return"]
    .rolling(window=3)
    .std()
)

df
```
