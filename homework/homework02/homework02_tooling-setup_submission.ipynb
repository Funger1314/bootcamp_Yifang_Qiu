## Completed Task 3 — Config Helper
Create helper functions to access keys and set up paths.


```python
import os
from pathlib import Path

PROJECT_ROOT = Path.cwd()

def get_config(name: str, default=None):
    return os.getenv(name, default)

def get_data_dir() -> Path:
    return Path(get_config("DATA_DIR", PROJECT_ROOT / "data"))

print("Project root:", PROJECT_ROOT)
print("Data dir:", get_data_dir())
```

## Completed Task 4 — Verify Keys and Paths Safely
Ensure `API_KEY` and `DATA_DIR` are available without printing secret values.


```python
api_key = get_config("API_KEY")
data_dir = get_data_dir()
data_dir.mkdir(parents=True, exist_ok=True)

print("API_KEY present:", bool(api_key))
print("DATA_DIR exists:", data_dir.exists())
print("DATA_DIR path:", data_dir.resolve())
```
